10. Get your selfie. Read it as image matrix. Intentionally add a noise matrix with your original image (Gaussian distribution, with mean = 110 and sigma = 5). Let us call the resultanant image as noisy image. Try to remove the noise by applying a (5x5) averaging filter/kernel. Let this be receovered image. Compare recovered image and the original image. Report the result in terms of MSE (mean square error)

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread('qn5.jpg', 0)

noise = np.random.normal(110, 5, image.shape)
noisy_image = image + noise

kernel = np.array([
    [1/9,1/9,1/9],
    [1/9,1/9,1/9],
    [1/9,1/9,1/9]
], dtype=np.float32)

recovered_image = cv2.filter2D(noisy_image, -1, kernel)

mse=np.mean((image - recovered_image)**2)
print(f"MSE: {mse}")

plt.figure(figsize=(10,10))

plt.subplot(1,3,1)
plt.imshow(image,cmap='gray')
plt.xticks([])
plt.yticks([])
plt.title('Original Image')

plt.subplot(1,3,2)
plt.imshow(noisy_image,cmap='gray')
plt.xticks([])
plt.yticks([])
plt.title('Noisy Image')

plt.subplot(1,3,3)
plt.imshow(recovered_image,cmap='gray')
plt.xticks([])
plt.yticks([])
plt.title('Recovered Image')

plt.show()
```

11. Take your selfie image. Use the rotational matrix operator, scaling matrix operator ans shear matrix operator on the image. List the challenges you face and record your results. DO NOT USE READYMADE functions/libraries/algorithms. You have to use the operartor matrices explicitly.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('qn11.jpg', 0)
h, w = img.shape

R = np.array([[0, -1], [1, 0]])       # Rotation: 90 degrees
S = np.array([[2, 0], [0, 2]])        # Scaling
H = np.array([[1, 0.5], [0, 1]])      # Shearing

def transform(A):
    out = np.zeros_like(img)
    # inv = np.linalg.inv(A)
    for y in range(h):
        for x in range(w):
            p = np.array([x-w//2, y-h//2])
            q = np.round(A @ p + [w//2, h//2]).astype(int)
            u, v = q

            if 0 <= u < w and 0 <= v < h:
                out[v, u] = img[y, x]

    return out

plt.figure(figsize=(10, 7))

plt.subplot(2, 2, 1)
plt.imshow(img, cmap='gray')
plt.title('Original')
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(transform(R), cmap='gray')
plt.title('Rotation')
plt.axis('off')

plt.subplot(2, 2, 3)
plt.imshow(transform(S), cmap='gray')
plt.title('Scaling')
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(transform(H), cmap='gray')
plt.title('Shearing')
plt.axis('off')

plt.tight_layout()
plt.show()
```

12. Record a video clip of (around 30 sec) a leaf using your mobile camera, with out shaking your hands. Track the pixel cordinates (rwo and column number) of the leaf apex from frame to frame. Now there are two time series signals, (i) row cooridnate as a function of frame number, and (ii) column cooridnate as a function of frame number. Apply 1D-FFT separetely on these signals, and apply 2D FFT.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

cap = cv2.VideoCapture('qn12.mp4')

fps = cap.get(cv2.CAP_PROP_FPS)
total = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
duration = total / fps

rows, cols = [], []
sample_frame = None

for t in np.arange(0, duration, 0.5):
    cap.set(cv2.CAP_PROP_POS_MSEC, t * 1000)
    ret, frame = cap.read()

    if not ret:
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    if sample_frame is None:
        sample_frame = gray

    rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

    fig, ax = plt.subplots()
    ax.imshow(rgb)
    ax.set_title(f'Time: {t:.1f}s - Click leaf apex')

    point = []

    def onclick(event):
        if event.inaxes == ax and event.xdata is not None:
            point.append((event.xdata, event.ydata))
            fig.canvas.mpl_disconnect(cid)
            plt.close(fig)

    cid = fig.canvas.mpl_connect('button_press_event', onclick)
    plt.show(block=False)

    while plt.fignum_exists(fig.number) and not point:
        plt.pause(0.1)

    if point:
        x, y = point[0]
        cols.append(round(x))
        rows.append(round(y))
    else:
        break

cap.release()
plt.close('all')

if len(rows) > 0:
    rows = np.array(rows)
    cols = np.array(cols)

    # 1D FFT
    R = np.fft.fft(rows - np.mean(rows))
    C = np.fft.fft(cols - np.mean(cols))

    # 2D FFT
    F = np.fft.fft2(sample_frame)
    spectrum = np.log(1 + np.abs(np.fft.fftshift(F)))

    # Display results
    plt.figure(figsize=(10, 8))

    plt.subplot(2, 2, 1)
    plt.plot(np.abs(R))
    plt.title('1D FFT - Row')

    plt.subplot(2, 2, 2)
    plt.plot(np.abs(C))
    plt.title('1D FFT - Column')

    plt.subplot(2, 2, 3)
    plt.imshow(sample_frame, cmap='gray')
    plt.title('Original Frame')
    plt.axis('off')

    plt.subplot(2, 2, 4)
    plt.imshow(spectrum, cmap='gray')
    plt.title('2D FFT Spectrum')
    plt.axis('off')

    plt.tight_layout()
    plt.show()
else:
    print("No points were selected.")
```

```python
import numpy as np
import matplotlib.pyplot as plt

image=np.zeros((100,100),dtype=np.uint8)
# Draw concentric triangles using polylines

cx,cy=50,50
for s in [10,20,30]:
    pts = np.array([[cx,cy-s],[cx-s,cy+s],[cx+s,cy+s]])
    cv2.polylines(image,[pts],True,1,1)
    # cv2.circle(image,(cx,cy),s,1,1)

    

# Rotate image by 45 degrees using a matrix
theta = np.radians(45)
R = np.array([[np.cos(theta), -np.sin(theta)],
              [np.sin(theta),  np.cos(theta)]])

rotated = np.zeros_like(image)
h, w = image.shape

for y in range(h):
    for x in range(w):
        p = np.array([x-w//2, y-h//2])
        u, v = np.round(np.linalg.inv(R) @ p + [w//2, h//2]).astype(int)

        if 0 <= u < w and 0 <= v < h:
            rotated[y, x] = image[v, u]

plt.figure(figsize=(8, 4))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap='gray')
plt.title('Concentric Triangles')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(rotated, cmap='gray')
plt.title('Rotated 45 Degrees')
plt.axis('off')

plt.show()
```
